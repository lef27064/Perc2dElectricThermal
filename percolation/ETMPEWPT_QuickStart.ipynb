Create necessary files, dirs & run a sample

```python
import os

# 1. Ensure we are in the percolation directory
if os.path.basename(os.getcwd()) != 'percolation':
    %cd /content/Perc2dElectricThermal/percolation

# 2. Create necessary directories
os.makedirs("inputs", exist_ok=True)
os.makedirs("outputs", exist_ok=True)

# 3. Create settings.txt
settings_content = """#Total Iterations to Run :
5
#Save image file(s) :
TRUE
#Save random image(s) :
1
#Save image(s) for specific iteration(s) :
FALSE
#Iterations to save :
1
#Current directory :
"./outputs/"
"""

with open("settings.txt", "w") as f:
    f.write(settings_content)

# 4. Create a test input file (inputs/example_2d.txt)
input_content = """#name
example_run
#total component
3
#components % weights
0.4 0.3 0.3
#Components type Geometry : RECTANGLE = 0, SLOPEDRECTANGLE = 1, ELLIPSE = 3, NONE = 4
4 3 1
#SizeType CONSTANT = 0, VARIABLE = 1, NONE = 3
3 0 0
#Special Weights
1.6532 1.3279 1.342
#Dimension X
10.0e-09 10.0e-09 1.0e-09
#Dimension Y
10.0e-09 10.0e-09 10.0e-09
#hoops
0.5e-09 0.5e-09 0.5e-09
#Electric Conductivities
1.e-16 1.0 0.5
#Thermal Conductivities
1.e-16 1.0 0.5
#Young Modulus
1.e+5 1.e+10 0.5e+9
#Poisson Ratio
0.3 0.32 0.31
#width - height
500 500
#iterations
5
#pixels per min Sphere
101
#Swiss chesse case TRUE or FALSE
FALSE
# CALC ELECTRIC CONDUCTIVITY
TRUE
# CALC ELECTRIC CONDUCTIVITY WITH FINITE DIFFERENCES
FALSE
# CALCULATE STATISTICS
FALSE
"""

with open("inputs/example_2d.txt", "w") as f:
    f.write(input_content)

print("Created settings.txt and inputs/example_2d.txt successfully!")

# 5. Εκτέλεση του ETMPEWPT
!./ETMPEWPT
```
